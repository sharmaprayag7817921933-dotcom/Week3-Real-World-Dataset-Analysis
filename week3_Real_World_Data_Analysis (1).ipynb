```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create a sample dataset
data = {
    "Student": ["Aman", "Priya", "Rahul", "Neha", "Rohan",
                "Simran", "Arjun", "Anjali", "Karan", "Pooja"],
    "Study_Hours": [2, 5, 3, 6, 4, 7, 1, 5, 3, 6],
    "Marks": [45, 85, 55, 90, 65, 95, 35, 80, 50, 88],
    "Attendance": [70, 95, 75, 98, 80, 99, 60, 90, 72, 96]
}

df = pd.DataFrame(data)

print("STUDENT PERFORMANCE DATASET")
display(df)
```

```python
print("1. DATASET INFORMATION")
df.info()

print("\n2. STATISTICAL SUMMARY")
display(df.describe())

print("\n3. MISSING VALUES")
print(df.isnull().sum())

print("\n4. DUPLICATE RECORDS")
print(df.duplicated().sum())
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Study_Hours"], df["Marks"])

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Student"], df["Marks"])

plt.title("Student Performance Analysis")
plt.xlabel("Student Name")
plt.ylabel("Marks")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Correlation analysis
correlation = df[["Study_Hours", "Marks", "Attendance"]].corr()

print("CORRELATION BETWEEN VARIABLES")
display(correlation)

# Correlation heatmap
plt.figure(figsize=(6, 4))
plt.imshow(correlation, cmap="coolwarm", vmin=-1, vmax=1)

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.colorbar(label="Correlation")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
# FINAL FINDINGS AND CONCLUSION

print("WEEK 3: REAL-WORLD DATASET ANALYSIS")
print("=" * 40)

print("\n1. DATASET OVERVIEW")
print("The dataset contains information about 10 students.")
print("It includes study hours, marks, and attendance.")

print("\n2. KEY FINDINGS")
print("Highest Marks:", df.loc[df["Marks"].idxmax(), "Student"],
      "-", df["Marks"].max())

print("Lowest Marks:", df.loc[df["Marks"].idxmin(), "Student"],
      "-", df["Marks"].min())

print("Average Marks:", round(df["Marks"].mean(), 2))
print("Average Study Hours:", round(df["Study_Hours"].mean(), 2))
print("Average Attendance:", round(df["Attendance"].mean(), 2))

print("\n3. DATA QUALITY")
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

print("\n4. CONCLUSION")
print("The sample data helps explore relationships between")
print("study hours, marks, and attendance.")
print("More real-world data is needed for reliable conclusions.")

print("\nPROJECT ANALYSIS COMPLETED!")
```
